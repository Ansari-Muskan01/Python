# If : 

>The if statement in Python is used to execute a block of code only if a specified condition is true.

>If the condition is false, the code inside the if block is skipped.

**Syntax :**

```python
if condition:
    print()
    
```

# if - elif - else

>The if-elif-else statement is used to check multiple conditions.

>If the first condition is true, it executes the code under if.

>If the first condition is false, it checks the next condition under elif.

>If none of the conditions are true, it executes the code under else.

**Syntax :**

```python
if condition:
    print()
elif condition :
    print()
else:
    print()
```

````python
# Write a Python program to check whether a given number is divisible by 3, divisible by 5, or divisible by both 3 and 5.
````

````python
# Condition1 :  divisible by 3
# Condition2 :  divisible by 5
# Condition3 :divisible by both 3 and 5

number = int(input("Enter the number: "))

# Condition1 :  divisible by 3
if number%3==0 and number%5==0:
    print("divisible by both 3 and 5")

# Condition2 :  divisible by 5
elif number%5==0:
    print("divisible by 5")

elif number%3==0 :
    print("divisible by 3")

else :
    print("Invalid Number")
````
