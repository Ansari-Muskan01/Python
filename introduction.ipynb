# Data Type

**Integer :** An integer stores whole numbers.

**Float :**  A float stores numbers with decimal points.

**Complex :** A complex number contains a real number and an imaginary number. In Python, the imaginary part uses j.

**String :** A string stores text, such as names, messages, and sentences. Strings can be enclosed in single quotes, double quotes, or triple quotes.

**Boolean :** A Boolean stores one of two values: True or False. The first letter must be uppercase.

```python
print(2+3j)
print(type(2+3j))
```

# Comments in Python

## 1. Single-Line Comment : 
>A single-line comment starts with the # symbol. Everything written after # on that line is treated as a comment.

**Example 1: Using #**

```python
# This is a multi-line comment.
# It explains the purpose of the program.
# Python ignores these comment lines.

print("Hello, Python!")
```

## 2. Multi-Line Comments

> Python does not have a separate syntax specifically for multi-line comments. We can write multiple single-line comments by starting each line with #. We can also use triple quotes (""" """ or ''' ''') to create multi-line strings, but they are not technically comments.

**Example 2: Using Triple Quotes**

```python
"""
This is a multi-line string.
It contains multiple lines of text.
"""

print("Hello, Python!")
```

# Escape characters in Python 

## 1. New Line (\n)
> The \n escape character moves the following text to a new line.

```python
print("Hello\nPython")
```

## 2. Tab Space (\t)
>The \t escape character adds a horizontal tab space between text.

```python
print("Hello              python")
```

```python
print("Hello\t\tPython")
```

## 3. Backslash (\\)
> The \\ escape character prints a single backslash.

```python
print("C:\\Python\\Programs")
```

## 4. Single Quotation Mark (\')
> The \' escape character allows a single quotation mark to appear inside a string enclosed in single quotes.

```python
print('It\'s Python')
```

## 5. Double Quotation Mark (\")
> The \" escape character allows double quotation marks inside a string enclosed in double quotes.

```python
print("He said, \"Hello!\"")
```
