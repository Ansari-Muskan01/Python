# Slicing and Indexing

```python
#       0123456789  -- Left to right
sent = 'thiS is PyThoN ProGmMing.'
#                          -3-2-1 
#right to left
```

```python
sent = 'thiS is PyThoN ProGmMing.'
sent[8]
```

```python
sent = 'thiS is PyThoN ProGmMing.'

# Positive Indexing
print("First character:", sent[0])
print("Fifth character:", sent[4])

# Negative Indexing
print("Last character:", sent[-1])
print("Second last character:", sent[-2])
```

```python
# Slicing
print("All characters:", sent[:]) # ALL
print("First five characters:", sent[:5]) # BEFORE
print("Afer:", sent[5:]) # After
print("Characters from index 5 to 10:", sent[5:11]) #STRT : END
print("Reverse string:", sent[::-1])#REVERSE
```

# Membership Operators in String 

```python
sent = 'thiS is PyThoN ProGmMing.'

# in operator
print("'Python' in string:", "Python" in sent)

# not in operator
print("'Java' not in string:", "Java" not in sent)
```

# Accessing string Elements Using Loops

```python
# for loop 
sent = 'thiS is PyThoN ProGmMing.'
for char in sent:
    print(char)
```

```python
sent = 'thiS is PyThoN ProGmMing.'

i = 0

while i < len(sent):
    print(sent[i])
    i += 1
```
