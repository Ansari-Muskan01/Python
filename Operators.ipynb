# Arithmetic Operators

> Arithmetic operators are used to perform mathematical operations such as addition, subtraction, multiplication, and division.

```python
a = 2
b = 3

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Floor Division:", a // b)
print("Modulus:", a % b)
print("Exponentiation:", a ** b)
```

## Assignment Operators
> Assignment operators are used to assign values to variables or update their existing values.

```python
a = 10
print("Assignment:", a)

a += 5
print("Addition Assignment:", a)

a -= 3
print("Subtraction Assignment:", a)

a *= 2
print("Multiplication Assignment:", a)

a /= 4
print("Division Assignment:", a)

a //= 3
print("Floor Division Assignment:", a)

a %= 3
print("Modulus Assignment:", a)

a **= 2
print("Exponentiation Assignment:", a)
```

# Comparison Operators
> Comparison operators are used to compare two values. They return either True or False.

```python
a = 10
b = 5

print("Equal to:", a == b)
print("Not equal to:", a != b)
print("Greater than:", a > b)
print("Less than:", a < b)
print("Greater than or equal to:", a >= b)
print("Less than or equal to:", a <= b)
```

# Logical Operators
> Logical operators are used to combine conditional expressions or reverse their results.

```python
T and T -- T
F and T -- F
T and F -- F
F and F -- F


T or T -- T
F or T -- T
T or F -- T
F or F -- F

not T -- F
not F -- T
```

```python
a = 10
b = 5
              #10>5  and 5>3
print("AND:", a > 5 and b > 3)
print("OR:", a < 5 or b > 3)   #10<5  and 5>3
print("NOT:", not(a > 5)) # 10 >5
```

# Membership Operators
> Membership operators are used to check whether a value exists in a sequence, such as a string, list, or tuple.

```python
fruits = ["Apple", "Banana", "Mango"]

print("Apple" in fruits)
print("Orange" not in fruits)
```

# Identity Operators

>Identity operators are used to check whether two variables refer to the same object in memory. They return either True or False.

```python
a = [10, 20, 30]
b = [10, 20, 30]

print(a is not b)
print(a is b)
```

```python
a = [10, 20, 30]
b = [10, 20, 30]

print(id(a))
print(id(b))


print(a is not b)
print(a is b)
```

```python
print(a is b)
print(a is not b)
```
