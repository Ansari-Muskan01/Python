# Range Function : 

The range() function in Python is used to generate a sequence of numbers. It is commonly used with for loops to execute a block of code a specific number of times.

**Syntax: range(start, stop, step)**

**Parameters:**

**- start**: The number from which the sequence starts (optional, default is 0).

**- stop:** The number at which the sequence stops (not included).

**- step:** The difference between each number (optional, default is 1).

````python
# Generate numbers from 0 to 4

range(4)  # range(stop) – End Position
````

````python
# Generate numbers from 1 to 5
range(1, 5)   #range(start, stop) – Start and End Position
````

````python
# Generate numbers from 1 to 4 with a step of 2
range(1, 4, 2)   #range(start, stop, step) – Start, End and Step
````

**Note :** 

The start value is optional. If not provided, it starts from 0.

The stop value is compulsory and is not included in the sequence.

The step value is optional. If not provided, it is 1.

The step value can be positive or negative, but it cannot be 0.

The range() function does not directly return a list. Use list() to convert it into a list.

We can also use a negative step to generate numbers in reverse order.

````python
# Convert the range of numbers into a list
list(range(1, 6))     #list(range()) into a List
````

````python
# Convert range into a tuple
tuple(range(1, 6))    #tuple(range()) into a List
````

````python
# Convert range into a set
set(range(1, 6))    #set(range()) into a List
````

````python
dict(range(1, 6))    # {Key : Value}
````

````python
# Convert range into a dictionary
dict.fromkeys(range(1, 6))
# Output: {1: None, 2: None, 3: None, 4: None, 5: None}
````

# 1 - factorial Number

##### 5! = 5 * 4 * 3 * 2 * 1 

````python
# for loop 
number = 5
factorial = 1 # 1  # 2 # 6 
for i in range( 1 , number+1 ):
    factorial = factorial * i # 2 * 3 
print(factorial)    
````

````python
# while loop 
number = 5 
factorial = 1
i = 0
while i < number:
    i += 1
    factorial *= i
print(factorial)
````

# 3 - Prime Number

````python
# for loop 
number = int(input("Enter the number: "))
count = 0 

for i in range(1,number+1):
    if number%i ==0:
        count = count + 1
print("Count: ",count)        
if count == 2 :
    print("Prime")
else :
    print("Not Prime")
````

# 7 - Fibonacci Series

````python
# while loop 
number = int(input("Enter the number: "))
previous = 0
current = 1 
print(previous)
print(current)

i = 1 
while i <= number :
    total = previous + current
    previous = current
    current = total
    i = i + 1 
    print(total)
````

````python
# for loop 
number = int(input("Enter the number : "))
previous = 0
current = 1
for i in range(1,number+1):
    total = previous + current
    previous = current
    current = total
    print(total)
````
