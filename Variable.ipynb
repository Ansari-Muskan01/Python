# Multiple Variable assignment

> In Python, we can assign values to multiple variables in a single line.

```python
# Example 1: Assigning different values to different variables

a, b, c = 10, 20, 30

print(a)
print(b)
print(c)
```

```python
print("Value of a: " ,a)
print("Value of b: ",b)
print("Value of c: ",c)
```

```python
# Example 2: Assigning the same value to multiple variables

a = b = c = 100

print(a)
print(b)
print(c)
```

# Variable swapping

> Variable swapping means exchanging the values of two variables. 

```python
# we can swap two variables without using a third variable.

a = 10
b = 20

# Swapping the values
a, b = b, a

print("Value of a:", a)
print("Value of b:", b)
```

# Type Casting in Python

```python
# int()
a = "10"
print("Before Type Casting:", a)
print("Before Data Type:", type(a))
print()
print("After Type Casting:", int(a))
print("After Data Type:", type(int(a)))
```

```python

# float()
b = "10.5"
print("Before Type Casting:", b)
print("Before Data Type:", type(b))
print()
print("After Type Casting:", float(b))
print("After Data Type:", type(float(b)))
```

```python
# str()
c = 100
print("Before Type Casting:", c)
print("Before Data Type:", type(c))
print()
print("After Type Casting:", str(c))
print("After Data Type:", type(str(c)))
```

```python
# bool()
d = 1
print("Before Type Casting:", d)
print("Before Data Type:", type(d))
print()
print("After Type Casting:", bool(d))
print("After Data Type:", type(bool(d)))
```

# Input Function 

> The input() function is used to accept information from the user.
By default, the input() function returns the entered value as a string.


```python
name = input("Enter your name: ")
print("Name: ", name)

print("Data Type:", type(name))
```

```python
Number = input("Enter the Number: ")
print("Number: ", Number)

print("Data Type:", type(Number))
```

```python
# note : The input() function accepts the user's input as a string. The int() function converts that input into an integer.
Number = int(input("Enter the Number: "))
print("Number:", Number)
print("Data Type:", type(Number))
```

```python
Number1 = int(input("Enter the Number: "))
print("Number:", Number1)
print("Data Type:", type(Number1))
```

```python
# Note: The int() function converts the user's input into an integer.
# It accepts whole numbers, such as 10, 25, or 100.
# If the user enters a decimal number like 10.5, it raises a ValueError.
```

```python
Number1 = float(input("Enter the Number: "))
print("Number:", Number1)
print("Data Type:", type(Number1))
```

```python
Number1 = float(input("Enter the Number: "))
print("Number:", Number1)
print("Data Type:", type(Number1))
```

```python
# Note: The float() function accepts whole numbers as well as decimal numbers and converts them into a float value. 
# For example, if the user enters 5, it converts the value to 5.0.
```

# eval()

>eval() : The eval() function is used to accept both integer and decimal values without manually converting them using int() or float(). It evaluates the entered value as a Python expression and returns the result with the appropriate data type.

```python
Number1 = eval(input("Enter the Number: "))
print("Number:", Number1)
print("Data Type:", type(Number1))
```

# Data Sturcture 

> Data structures in Python are used to store and organize data so that it can be accessed and managed efficiently.

**Types of Data Structures in Python:**

**List  –** Used to store multiple items in an ordered, changeable collection.

**Tuple  –** Used to store multiple items in an ordered, unchangeable collection.

**Set  –** Used to store unique items without duplicates.

**Dictionary  –** Used to store data in key-value pairs.

**String  –** Used to store text.
