# Slicing and Indexing

```python
numbers = (10, 20, 30, 40, 50)

# Access the first element
print(numbers[0])

# Access the third element
print(numbers[2])

# Access the last element
print(numbers[-1])
```

```python
numbers = (10, 20, 30, 40, 50)

# Access all values
print("All elements:", numbers[:])

# Access elements from index 1 to 3
print("Index 1 to 3:", numbers[1:4])

# Access the first three elements
print("Before:", numbers[:3])

# Access elements from index 2 to the end
print("After:", numbers[2:])

# Reverse the tuple
print("Reverse:", numbers[::-1])
```

## Tuple Operations

```python
T1 = (10, 20, 30)
T2 = (40, 50, 60)

T3 = T1 + T2
print(T3)
```

## Membership Operators in Tuple

```python
a = (10, 20, 30, 40, 50, 60)
print(10 in a)
print(10 not in a)
```

## Accessing Tuple Elements Using Loops

```python
# for loop:

T1 = (10, 20, 30, 40, 50)
for i in T1:
    print(i)
```

```python
len(T1)
```

```python
# while loop:

T1 = (10, 20, 30, 40, 50)
i = 0
while i < len(T1):
    print(T1[i]) #T1 = (10, 20, 30, 40, 50)[i]#0
    i += 1
```

## Basic Built-in Functions in Tuple

```python
T1 = (10, 20, 30, 40, 50)

# len()
print("Total number of elements:", len(T1))

# max()
print("Largest element:", max(T1))

# min()
print("Smallest element:", min(T1))

# sum()
print("Sum of all elements:", sum(T1))
```
